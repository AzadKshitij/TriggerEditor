```python
import duckdb
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../check.csv')
```

```python
df
```

```python
duck = duckdb.connect(':memory:')
duck.register('df', df)
```

```python
query = """CASE 
    WHEN [Difficulty] > 7 THEN 'High'
    WHEN [Difficulty] > 6 THEN 'Medium'
    ELSE 'Low'
END"""
new_column = 'Category'
```

```python
for col in df.columns:
    query = query.replace(f'[{col}]', f'{col}')
```

```python
result = duck.execute(f'SELECT *, {query} as {new_column} FROM df').fetchdf()
print(result)
```

```python
result = duck.execute(f'SELECT CURRENT_TIME as Today FROM df').fetchdf()
result
```
